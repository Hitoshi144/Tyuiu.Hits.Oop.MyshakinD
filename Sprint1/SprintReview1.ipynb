```polyglot-notebook
public class Product {
    private string Name { get; set; }
    private decimal Price { get; set; }
    private int Quantity { get; set; }
    protected string Category { get; set; }
    internal string Description { get; set; }

    public Product(string name, decimal price, int quantity, string category, string description) {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice() {
        return Quantity * Price;
    }

    public void UpdateQuantity(int amount) {
        Quantity += amount;
    }

    public string GetProductInfo() {
        return $"Имя товара: {Name}\nОписание: {Description}\nКатегория: {Category}\nЦена: {Price}\nКоличество: {Quantity}";
    }

    public int GetQuantity() => Quantity;
    public void SetQuantity(int quantity) => Quantity = quantity;
}

public class Seller {
    private string Name { get; set; }
    private string EmployeeId { get; set; }
    protected decimal Salary { get; set; }
    internal string ContactInfo { get; set; }

    private List<Product> Products { get; set; }

    public Seller(string name, string id, decimal salary, string contact) {
        Name = name;
        EmployeeId = id;
        Salary = salary;
        ContactInfo = contact;
        Products = new List<Product>();
    }

    public void AddProduct(Product product, int quantity) {
        var existingProduct = Products.FirstOrDefault(p => p.GetProductInfo() == product.GetProductInfo());
        
        if (existingProduct != null) {
            // Если товар уже есть, увеличиваем его количество
            existingProduct.UpdateQuantity(quantity);
        } else {
            // Если товара нет, добавляем новый
            var newProduct = new Product(product.GetProductInfo().Split('\n')[0].Replace("Имя товара: ", ""), 
                                       decimal.Parse(product.GetProductInfo().Split('\n')[3].Replace("Цена: ", "")), 
                                       quantity,
                                       product.GetProductInfo().Split('\n')[2].Replace("Категория: ", ""),
                                       product.GetProductInfo().Split('\n')[1].Replace("Описание: ", ""));
            Products.Add(newProduct);
        }

        Console.WriteLine($"Кол-во товаров у продавца: {GetTotalProductCount()}\nРазных товаров на складе: {Products.Count}");
    }

    public void SellProduct(string productName, int quantity) {
        var product = Products.FirstOrDefault(p => p.GetProductInfo().Contains(productName));
        
        if (product != null && product.GetQuantity() >= quantity) {
            product.UpdateQuantity(-quantity);
            
            if (product.GetQuantity() <= 0) {
                Products.Remove(product);
            }
            
            Console.WriteLine($"Продано {quantity} единиц товара {productName}");
            Console.WriteLine($"Остаток товаров: {GetTotalProductCount()}\nРазных товаров: {Products.Count}");
        } else {
            Console.WriteLine("Недостаточно товара или товар не найден");
        }
    }

    private int GetTotalProductCount() {
        return Products.Sum(p => p.GetQuantity());
    }

    public string GetSellerInfo() {
        return $"Имя: {Name}\nId: {EmployeeId}\nЗарплата: {Salary}\nКонтакты: {ContactInfo}\nВсего товаров: {GetTotalProductCount()}\nРазных товаров: {Products.Count}";
    }
}

public class Store {
    private string StoreName { get; set; }
    private string Location { get; set; }
    public string StoreHours { get; set; }

    private List<Seller> Sellers { get; set; }

    public Store(string storename, string location, string storehours) {
        StoreName = storename;
        Location = location;
        StoreHours = storehours;
        Sellers = new List<Seller>();
    }

    public void AddSeller(Seller seller) {
        Sellers.Add(seller);
        Console.WriteLine($"Продавец добавлен. Всего продавцов: {Sellers.Count}");
    }

    public void ListProducts() {
        Console.WriteLine($"\n--- Информация о всех продавцах магазина '{StoreName}' ---");
        foreach (var seller in Sellers) {
            Console.WriteLine(seller.GetSellerInfo());
            Console.WriteLine("---");
        }
    }

    public string GetStoreInfo() {
        return $"Имя магазина: {StoreName}\nМестоположение: {Location}\nЧасы работы: {StoreHours}\nКоличество поставщиков: {Sellers.Count}";
    }
}

Product newProduct = new Product("Молоко", 58, 5, "Молочка", "Жирность 3.2%");
Console.WriteLine("=== ИНФОРМАЦИЯ О ТОВАРЕ ===");
Console.WriteLine(newProduct.GetProductInfo());
Console.WriteLine($"Общая цена: {newProduct.GetTotalPrice()}");
newProduct.UpdateQuantity(3);
Console.WriteLine($"Общая цена после обновления: {newProduct.GetTotalPrice()}");

Console.WriteLine("\n=== РАБОТА ПРОДАВЦА ===");
Seller newSeller = new Seller("Барыга", "13fs", 6000, "+77777777777");
Console.WriteLine(newSeller.GetSellerInfo());
newSeller.AddProduct(newProduct, 3);
newSeller.SellProduct("Молоко", 1);

Console.WriteLine("\n=== РАБОТА МАГАЗИНА ===");
Store newStore = new Store("Пяторка", "Мельнкайте, гдето там", "12:00-13:00");
Console.WriteLine(newStore.GetStoreInfo());
newStore.AddSeller(newSeller);
newStore.ListProducts();
```
