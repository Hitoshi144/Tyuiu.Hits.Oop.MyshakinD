<h1 style="color:DodgerBlue">Явная реализация интерфейса</h1>

В C# интерфейсы могут быть реализованы явным и неявным образом. Давайте подробно рассмотрим, что такое явная реализация интерфейсов, реализация интерфейсов в базовых и производных классах, а также наследование интерфейсов, используя примеры классов `Vehicle`, `Car` и `Motorcycle`.

### Явная реализация интерфейсов

Явная реализация интерфейса используется для того, чтобы скрыть методы интерфейса от стандартного контекста класса. Это означает, что методы интерфейса могут быть вызваны только через ссылку на интерфейс, а не через экземпляр класса.

#### Пример явной реализации

Рассмотрим интерфейс `IVehicle` и его явную реализацию в классе `Car`.

```csharp
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    // Явная реализация интерфейса
    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    // Неявная реализация метода
    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

В этом примере методы `Start` и `Stop` реализованы явно, и для их вызова необходимо использовать объект типа `IVehicle`:

```csharp

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start(); // Корректно, вызывает явную реализацию
// myCar.Stop(); // Корректно, вызывает явную реализацию
```

Если вы попытаетесь вызвать `Start` или `Stop` непосредственно через экземпляр `Car`, это приведёт к ошибке компиляции:

```csharp
Car myCar = new Car();
myCar.Start(); // Ошибка компиляции
```

### Реализация интерфейсов в базовых и производных классах

Интерфейсы могут быть реализованы в базовых классах и наследоваться в производных классах. Давайте рассмотрим эту концепцию на примере, включающем классы `Vehicle`, `Car` и `Motorcycle`.

#### Базовый класс Vehicle

```csharp
public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();  // Оставляем абстрактным
    public abstract void Stop();   // Оставляем абстрактным

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}
```

#### Реализация в производных классах

Теперь реализуем `Car` и `Motorcycle`.

```csharp
public class Car : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }
}

public class Motorcycle : Vehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }
}
```

Теперь оба класса `Car` и `Motorcycle` обязаны реализовать методы `Start` и `Stop`, которые определены в абстрактном классе `Vehicle`.

### Наследование интерфейсов

Интерфейсы могут наследовать друг от друга. Это позволяет создавать сложные иерархии интерфейсов. Например, мы можем создать интерфейс `IMotorizedVehicle`, который наследует `IVehicle`.

```csharp
public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}
```

Таким образом, `Motorcycle` теперь может реализовать все методы из обоих интерфейсов `IVehicle` и `IMotorizedVehicle`.

### Заключение

1. **Явная реализация** интерфейсов скрывает функциональность классов от общего доступа, что помогает избежать путаницы с методами.
   
2. **Реализация интерфейсов в базовых и производных классах** позволяет создавать иерархию классов и обеспечивает обязательность реализации методов в производных классах.

3. **Наследование интерфейсов** позволяет комбинировать несколько интерфейсов, создавая более сложные структуры, что делает код более гибким и расширяемым.

Эти механизмы являются важными инструментами в инструменте разработчика, позволяя создавать более структурированные и управляемые программы.

````
public interface IVehicle
{
    void Start();
    void Stop();
    int GetCurrentSpeed();
}

public class Car : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    private int currentSpeed;

    void IVehicle.Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} запущен.");
    }

    void IVehicle.Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} остановлен.");
    }

    public int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

IVehicle myCar = new Car { Make = "Toyota", Model = "Camry" };
myCar.Start();

public abstract class Vehicle : IVehicle
{
    public string Make { get; set; }
    public string Model { get; set; }
    protected int currentSpeed;

    public abstract void Start();
    public abstract void Stop();

    public virtual int GetCurrentSpeed()
    {
        return currentSpeed;
    }
}

public interface IMotorizedVehicle : IVehicle
{
    void RevEngine();
}

public class Motorcycle : Vehicle, IMotorizedVehicle
{
    public override void Start()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл запущен.");
    }

    public override void Stop()
    {
        currentSpeed = 0;
        Console.WriteLine($"{Make} {Model} мотоцикл остановлен.");
    }

    public void RevEngine()
    {
        Console.WriteLine("Мотоцикл рычит!");
    }
}

Motorcycle myMoto = new Motorcycle();
myMoto.Start();
myMoto.RevEngine();
myMoto.Stop();
````

Ниже в блоке по примеру создайте базовый класс Animal и производные классы (3-4 например Dog, Cat, Bird и так далее) реализуйте структуру и объявление класса, включая свойства, геттеры и сеттеры, а также применить явнаю реализацию интерфейса.

````
public interface IAnimal
{
    void MakeSound();
    void Move();
    string GetInfo();
}

public abstract class Animal : IAnimal
{
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    
    private string _species;
    public string Species 
    { 
        get => _species; 
        protected set => _species = value; 
    }
    
    protected bool isSleeping;

    public Animal(string name, int age, double weight, string species)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Species = species;
        isSleeping = false;
    }

    public abstract void MakeSound();
    public abstract void Move();

    public virtual string GetInfo()
    {
        return $"{Name} ({Species}), Возраст: {Age} лет, Вес: {Weight} кг";
    }

    public void Sleep()
    {
        isSleeping = true;
        Console.WriteLine($"{Name} спит.");
    }

    public void WakeUp()
    {
        isSleeping = false;
        Console.WriteLine($"{Name} проснулся.");
    }
}

public class Dog : Animal, IAnimal
{
    public string Breed { get; set; }
    private bool isTailWagging;

    public Dog(string name, int age, double weight, string breed) 
        : base(name, age, weight, "Собака")
    {
        Breed = breed;
        isTailWagging = false;
    }

    void IAnimal.MakeSound()
    {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} тихо посапывает во сне...");
        }
        else
        {
            Console.WriteLine($"{Name} громко лает: Гав-гав!");
        }
    }

    void IAnimal.Move()
    {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} ворочается во сне.");
        }
        else
        {
            Console.WriteLine($"{Name} бегает и виляет хвостом.");
            isTailWagging = true;
        }
    }

    string IAnimal.GetInfo()
    {
        return base.GetInfo() + $", Порода: {Breed}";
    }

    public void Fetch()
    {
        if (!isSleeping)
        {
            Console.WriteLine($"{Name} приносит палку!");
        }
    }

    public void WagTail()
    {
        isTailWagging = true;
        Console.WriteLine($"{Name} виляет хвостом.");
    }

    public override void MakeSound() {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} тихо посапывает во сне...");
        }
        else
        {
            Console.WriteLine($"{Name} громко лает: Гав-гав!");
        }
    }

    public override void Move() {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} ворочается во сне.");
        }
        else
        {
            Console.WriteLine($"{Name} бегает и виляет хвостом.");
            isTailWagging = true;
        }
    }
}

public class Cat : Animal, IAnimal
{
    public bool HasClaws { get; set; }
    private int miceCaught;

    public Cat(string name, int age, double weight, bool hasClaws = true) 
        : base(name, age, weight, "Кошка")
    {
        HasClaws = hasClaws;
        miceCaught = 0;
    }

    void IAnimal.MakeSound()
    {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} мурлычет во сне...");
        }
        else
        {
            Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
        }
    }

    void IAnimal.Move()
    {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} потягивается во сне.");
        }
        else
        {
            Console.WriteLine($"{Name} грациозно крадется.");
        }
    }

    string IAnimal.GetInfo()
    {
        string clawsInfo = HasClaws ? "с когтями" : "без когтей";
        return base.GetInfo() + $", {clawsInfo}";
    }

    public override void MakeSound() {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} мурлычет во сне...");
        }
        else
        {
            Console.WriteLine($"{Name} мяукает: Мяу-мяу!");
        }
    }

    public override void Move() {
        if (isSleeping)
        {
            Console.WriteLine($"{Name} потягивается во сне.");
        }
        else
        {
            Console.WriteLine($"{Name} грациозно крадется.");
        }
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} мурлычет: Мррррр...");
    }

    public void CatchMouse()
    {
        miceCaught++;
        Console.WriteLine($"{Name} поймал мышь! Всего поймано: {miceCaught}");
    }
}

public class Bird : Animal, IAnimal
{
    public double Wingspan { get; set; }
    public bool CanFly { get; set; }
    private double flightAltitude;

    public Bird(string name, int age, double weight, double wingspan, bool canFly = true) 
        : base(name, age, weight, "Птица")
    {
        Wingspan = wingspan;
        CanFly = canFly;
        flightAltitude = 0;
    }

    void IAnimal.MakeSound()
    {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    void IAnimal.Move()
    {
        if (CanFly && !isSleeping)
        {
            flightAltitude = 10;
            Console.WriteLine($"{Name} летит на высоте {flightAltitude} метров.");
        }
        else if (!isSleeping)
        {
            Console.WriteLine($"{Name} прыгает по земле.");
        }
        else
        {
            Console.WriteLine($"{Name} спит на ветке.");
        }
    }

    string IAnimal.GetInfo()
    {
        string flyInfo = CanFly ? "летающая" : "нелетающая";
        return base.GetInfo() + $", Размах крыльев: {Wingspan} см, {flyInfo}";
    }

    public override void MakeSound() {
        Console.WriteLine($"{Name} чирикает: Чик-чирик!");
    }

    public override void Move() {
        if (CanFly && !isSleeping)
        {
            flightAltitude = 10;
            Console.WriteLine($"{Name} летит на высоте {flightAltitude} метров.");
        }
        else if (!isSleeping)
        {
            Console.WriteLine($"{Name} прыгает по земле.");
        }
        else
        {
            Console.WriteLine($"{Name} спит на ветке.");
        }
    }

    public void FlyTo(double altitude)
    {
        if (CanFly && !isSleeping)
        {
            flightAltitude = altitude;
            Console.WriteLine($"{Name} взлетел на высоту {flightAltitude} метров.");
        }
    }

    public void BuildNest()
    {
        Console.WriteLine($"{Name} строит гнездо.");
    }
}

public class Fish : Animal, IAnimal
{
    public string WaterType { get; set; } // "пресная" или "соленая"
    private double depth;

    public Fish(string name, int age, double weight, string waterType = "пресная") 
        : base(name, age, weight, "Рыба")
    {
        WaterType = waterType;
        depth = 0;
    }

    void IAnimal.MakeSound()
    {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }

    void IAnimal.Move()
    {
        if (!isSleeping)
        {
            depth += 1;
            Console.WriteLine($"{Name} плавает на глубине {depth} метров.");
        }
        else
        {
            Console.WriteLine($"{Name} неподвижно плавает во сне.");
        }
    }

    string IAnimal.GetInfo()
    {
        return base.GetInfo() + $", Тип воды: {WaterType}";
    }

    public override void MakeSound() {
        Console.WriteLine($"{Name} пускает пузыри: Буль-буль!");
    }

    public override void Move() {
        if (!isSleeping)
        {
            depth += 1;
            Console.WriteLine($"{Name} плавает на глубине {depth} метров.");
        }
        else
        {
            Console.WriteLine($"{Name} неподвижно плавает во сне.");
        }
    }

    public void DiveTo(double newDepth)
    {
        depth = newDepth;
        Console.WriteLine($"{Name} нырнул на глубину {depth} метров.");
    }

    public void BlowBubbles()
    {
        Console.WriteLine($"{Name} пускает пузыри!");
    }
}

Dog dog = new Dog("Бобик", 3, 15.5, "Лабрадор");
Cat cat = new Cat("Мурка", 2, 4.2, true);
Bird bird = new Bird("Кеша", 1, 0.3, 25.0, true);
Fish fish = new Fish("Немо", 1, 0.5, "соленая");

IAnimal[] animals = { dog, cat, bird, fish };

Console.WriteLine("=== Демонстрация явной реализации интерфейса ===");
foreach (IAnimal animal in animals)
{
    Console.WriteLine($"\n--- {animal.GetInfo()} ---");
    animal.MakeSound();
    animal.Move();
}

Console.WriteLine("\n=== Демонстрация собственных методов ===");
        
dog.WagTail();
dog.Fetch();
((IAnimal)dog).MakeSound();
        
cat.Purr();
cat.CatchMouse();
cat.Sleep();
((IAnimal)cat).MakeSound();
        
bird.FlyTo(15.5);
bird.BuildNest();
        
fish.DiveTo(5.0);
fish.BlowBubbles();
````
