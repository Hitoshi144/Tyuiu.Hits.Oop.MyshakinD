Создать базовый класс Item в C#, который будет представлять информацию о
товарах, которые могут быть заказаны или возвращены. На основе этого класса
разработать 2-3 производных класса, демонстрирующих принципы наследования и
полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и
методы, а также переопределены некоторые методы базового класса для
демонстрации полиморфизма.

Требования к базовому классу Item:
* Атрибуты: ID товара (ItemId), Название (Name), Цена (Price).
* Методы:
  * GetDetails(): метод для получения детальной информации о товаре.
  * CalculateDiscount(): метод для расчета скидки на товар.
  * ApplyDiscount(decimal discount): метод для применения скидки к цене
товара.

Требования к производным классам:
1. ЕдиничныйТовар (SingleItem): Должен содержать дополнительные атрибуты,
такие как Единица измерения (UnitMeasure). Метод GetDetails() должен быть
переопределен для добавления информации о единице измерения товара.
2. ПакетныйТовар (PackageItem): Должен содержать дополнительные
атрибуты, такие как Количество единиц в пакете (QuantityPerPackage).
Метод CalculateDiscount() должен быть переопределен для учета количества
единиц в пакете при расчете скидки.
3. СпециальныйТовар (SpecialItem) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Дата истечения скидки
(DiscountExpirationDate). Метод ApplyDiscount() должен быть переопределен
для добавления информации о сроке действия скидки.

```
public class ItemManager<T> where T : Item
{
    private List<T> items = new List<T>();

    public void AddItem(T item)
    {
        items.Add(item);
        Console.WriteLine($"Товар '{item.Name}' добавлен в менеджер.");
    }

    public void RemoveItem(int itemId)
    {
        var item = items.FirstOrDefault(i => i.ItemId == itemId);
        if (item != null)
        {
            items.Remove(item);
            Console.WriteLine($"Товар '{item.Name}' удален из менеджера.");
        }
    }

    public void DisplayAllItems()
    {
        Console.WriteLine($"\n--- Все товары в менеджере ({typeof(T).Name}) ---");
        foreach (var item in items)
        {
            item.GetDetails();
        }
    }

    public T FindItemById(int itemId)
    {
        return items.FirstOrDefault(i => i.ItemId == itemId);
    }

    public decimal CalculateTotalValue()
    {
        return items.Sum(item => item.Price);
    }

    public void ProcessItems<U>(Func<T, U> processor)
    {
        foreach (var item in items)
        {
            var result = processor(item);
            Console.WriteLine($"Обработан товар: {item.Name}, результат: {result}");
        }
    }
}

public class Inventory<T> where T : Item
{
    private Dictionary<int, int> stock = new Dictionary<int, int>();

    public void AddToStock(T item, int quantity)
    {
        if (stock.ContainsKey(item.ItemId))
            stock[item.ItemId] += quantity;
        else
            stock[item.ItemId] = quantity;

        Console.WriteLine($"Добавлено {quantity} единиц товара '{item.Name}' на склад. Текущий запас: {stock[item.ItemId]}");
    }

    public bool RemoveFromStock(T item, int quantity)
    {
        if (stock.ContainsKey(item.ItemId) && stock[item.ItemId] >= quantity)
        {
            stock[item.ItemId] -= quantity;
            Console.WriteLine($"Изъято {quantity} единиц товара '{item.Name}' со склада. Остаток: {stock[item.ItemId]}");
            return true;
        }
        Console.WriteLine($"Недостаточно товара '{item.Name}' на складе");
        return false;
    }

    public int GetStockQuantity(T item)
    {
        return stock.ContainsKey(item.ItemId) ? stock[item.ItemId] : 0;
    }
}

public class Item
{
    public int ItemId { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Category { get; set; }
    public string Manufacturer { get; set; }
    public double Weight { get; set; }

    public string Barcode { get; set; }
    public DateTime CreatedDate { get; set; }
    public bool IsAvailable { get; set; }
    public int Rating { get; set; }

    public Item(int itemId, string name, decimal price)
    {
        ItemId = itemId;
        Name = name;
        Price = price;
        CreatedDate = DateTime.Now;
        IsAvailable = true;
    }

    public Item() 
    {
        CreatedDate = DateTime.Now;
        IsAvailable = true;
    }

    public virtual void GetDetails()
    {
        Console.WriteLine($"ItemId: {ItemId}");
        Console.WriteLine($"Item Name: {Name}");
        Console.WriteLine($"Item Price: {Price}");
        Console.WriteLine($"Category: {Category}");
        Console.WriteLine($"Manufacturer: {Manufacturer}");
        Console.WriteLine($"Weight: {Weight} кг");
        Console.WriteLine($"Barcode: {Barcode}");
        Console.WriteLine($"Created: {CreatedDate.ToShortDateString()}");
        Console.WriteLine($"Available: {IsAvailable}");
        Console.WriteLine($"Rating: {new string('★', Rating)}{new string('☆', 5 - Rating)}\n");
    }

    public virtual void CalculateDiscount(decimal discount)
    {
        Console.WriteLine($"Price with discount = {Price - (Price * discount)}\n");
    }

    public virtual void CalculateDiscount(decimal discount, string discountName)
    {
        decimal discountedPrice = Price - (Price * discount);
        Console.WriteLine($"Price with {discountName} discount = {discountedPrice}\n");
    }

    public virtual void ApplyDiscount(decimal discount)
    {
        Price -= Price * discount;
    }

    public virtual void UpdateRating(int newRating)
    {
        Rating = Math.Clamp(newRating, 0, 5);
        Console.WriteLine($"Рейтинг товара '{Name}' обновлен: {Rating}");
    }

    public virtual string GetItemType()
    {
        return "Базовый товар";
    }

    public void ChangePrice(decimal newPrice)
    {
        Price = newPrice;
        Console.WriteLine($"Цена товара '{Name}' изменена на {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} (ID: {ItemId}) - {Price} руб.";
    }
}

public interface IPerishable
{
    DateTime ExpirationDate { get; set; }
    void CheckExpiration();
    bool IsExpired();
}

public interface IShippable
{
    string DeliveryMethod { get; set; }
    void Ship();
    decimal CalculateShippingCost();
}

public class SingleItem : Item
{
    public string UnitMeasure { get; set; }
    public bool IsOrganic { get; set; }
    public string OriginCountry { get; set; }

    public string QualityGrade { get; set; }
    public double Volume { get; set; }
    public bool IsRecyclable { get; set; }

    public SingleItem(int itemId, string name, decimal price, string unitMeasure)
        : base(itemId, name, price)
    {
        UnitMeasure = unitMeasure;
    }

    public SingleItem() {}

    public override void GetDetails()
    {
        base.GetDetails();
        Console.WriteLine($"Item UnitMeasure: {UnitMeasure}");
        Console.WriteLine($"Organic: {IsOrganic}");
        Console.WriteLine($"Origin Country: {OriginCountry}");
        Console.WriteLine($"Quality Grade: {QualityGrade}");
        Console.WriteLine($"Volume: {Volume}");
        Console.WriteLine($"Recyclable: {IsRecyclable}\n");
    }

    public override string GetItemType()
    {
        return "Штучный товар";
    }

    public void CheckQuality()
    {
        Console.WriteLine($"Качество товара '{Name}': {QualityGrade}");
    }

    public void Recycle()
    {
        if (IsRecyclable)
            Console.WriteLine($"Товар '{Name}' отправлен на переработку.");
        else
            Console.WriteLine($"Товар '{Name}' не подлежит переработке.");
    }

    public void ShowLabel()
    {
        Console.WriteLine($"Этикетка: {Name}, {Price} руб. за {UnitMeasure}");
    }

    public void ShowLabel(bool includeOrigin)
    {
        if (includeOrigin)
            Console.WriteLine($"Этикетка: {Name}, {Price} руб. за {UnitMeasure}, Страна: {OriginCountry}");
        else
            ShowLabel();
    }
}

public class PackageItem : Item, IShippable
{
    public int QuantityPerPackage { get; set; }
    public string DeliveryMethod { get; set; }
    public double PackageWeight { get; set; }
    public bool IsFragile { get; set; }

    public string PackageType { get; set; }
    public int ItemsInStock { get; set; }
    public bool IsBundle { get; set; }

    public PackageItem(int itemId, string name, decimal price, int quantityPerPackage)
        : base(itemId, name, price)
    {
        QuantityPerPackage = quantityPerPackage;
    }

    public PackageItem() {}

    public override void CalculateDiscount(decimal discount)
    {
        decimal totalDiscount = (Price * discount) * QuantityPerPackage;
        Console.WriteLine($"Package Discount = {totalDiscount}\n");
    }

    public override void CalculateDiscount(decimal discount, string discountName)
    {
        decimal totalDiscount = (Price * discount) * QuantityPerPackage;
        decimal totalPrice = Price * QuantityPerPackage;
        Console.WriteLine($"Total {discountName} discount for package = {totalDiscount}");
        Console.WriteLine($"Final price = {totalPrice - totalDiscount}\n");
    }

    public void Ship()
    {
        Console.WriteLine($"Товар '{Name}' отправлен методом доставки: {DeliveryMethod}");
    }

    public decimal CalculateShippingCost()
    {
        decimal baseCost = 100m;
        decimal fragileCost = IsFragile ? 50m : 0m;
        decimal weightCost = (decimal)PackageWeight * 10m;
        
        return baseCost + fragileCost + weightCost;
    }

    public void CheckPackaging()
    {
        Console.WriteLine(IsFragile
            ? "Требуется аккуратная упаковка."
            : "Стандартная упаковка подходит.");
    }

    public void Restock(int quantity)
    {
        ItemsInStock += quantity;
        Console.WriteLine($"Пополнен запас товара '{Name}' на {quantity} единиц. Текущий запас: {ItemsInStock}");
    }

    public override string GetItemType()
    {
        return "Упакованный товар";
    }
}

public class SpecialItem : Item, IPerishable, IShippable
{
    public DateTime DiscountExpirationDate { get; set; }
    public DateTime ExpirationDate { get; set; }
    public string DeliveryMethod { get; set; }
    public bool RequiresCooling { get; set; }
    public string StorageCondition { get; set; }

    public string SpecialCategory { get; set; }
    public int LoyaltyPoints { get; set; }
    public bool IsLimitedEdition { get; set; }

    public SpecialItem(int itemId, string name, decimal price, DateTime discountExpirationDate)
        : base(itemId, name, price)
    {
        DiscountExpirationDate = discountExpirationDate;
    }

    public SpecialItem() {}

    public override void ApplyDiscount(decimal discount)
    {
        if (DateTime.Now <= DiscountExpirationDate)
        {
            decimal oldPrice = Price;
            base.ApplyDiscount(discount);
            Console.WriteLine($"Скидка {discount:P} применена до {DiscountExpirationDate.ToShortDateString()}");
            Console.WriteLine($"Старая цена: {oldPrice}, Новая цена: {Price}\n");
        }
        else
        {
            Console.WriteLine("Срок действия скидки истек. Скидка не применена.\n");
        }
    }

    public void ApplyDiscount(decimal discount, bool forceApply)
    {
        if (forceApply || DateTime.Now <= DiscountExpirationDate)
        {
            decimal oldPrice = Price;
            base.ApplyDiscount(discount);
            Console.WriteLine($"Скидка {(forceApply ? "принудительно " : "")}применена");
            Console.WriteLine($"Старая цена: {oldPrice}, Новая цена: {Price}\n");
        }
        else
        {
            Console.WriteLine("Срок действия скидки истек. Скидка не применена.\n");
        }
    }

    public void CheckExpiration()
    {
        if (DateTime.Now > ExpirationDate)
            Console.WriteLine($"Товар '{Name}' просрочен!");
        else
            Console.WriteLine($"Товар '{Name}' годен до {ExpirationDate.ToShortDateString()}.");
    }

    public bool IsExpired()
    {
        return DateTime.Now > ExpirationDate;
    }

    public void Ship()
    {
        Console.WriteLine($"'{Name}' отправлен через {DeliveryMethod}. Условия хранения: {StorageCondition}");
    }

    public decimal CalculateShippingCost()
    {
        decimal baseCost = 150m;
        decimal coolingCost = RequiresCooling ? 100m : 0m;
        decimal urgentCost = IsLimitedEdition ? 200m : 0m;
        
        return baseCost + coolingCost + urgentCost;
    }

    public void KeepCool()
    {
        if (RequiresCooling)
            Console.WriteLine($"'{Name}' хранится в холодильнике.");
        else
            Console.WriteLine($"'{Name}' не требует охлаждения.");
    }

    public void AwardLoyaltyPoints(int points)
    {
        LoyaltyPoints += points;
        Console.WriteLine($"Начислено {points} бонусных баллов за покупку '{Name}'. Всего баллов: {LoyaltyPoints}");
    }

    public override string GetItemType()
    {
        return "Специальный товар" + (IsLimitedEdition ? " (Ограниченный тираж)" : "");
    }
}

public class ShoppingCart
{
    private List<Item> items = new List<Item>();

    public void AddItem(Item item)
    {
        items.Add(item);
        Console.WriteLine($"Товар {item.Name} добавлен в корзину.");
    }

    public void ShowCart()
    {
        Console.WriteLine("\n--- Корзина ---");
        foreach (var item in items)
        {
            item.GetDetails();
        }
    }

    public void CalculateTotal()
    {
        decimal total = 0;
        foreach (var item in items)
        {
            total += item.Price;
        }
        Console.WriteLine($"Итого к оплате: {total} руб.\n");
    }

    public void DisplayItemTypes()
    {
        Console.WriteLine("\n--- Типы товаров в корзине ---");
        foreach (var item in items)
        {
            Console.WriteLine($"{item.Name}: {item.GetItemType()}");
        }
    }
}

        SingleItem single = new SingleItem(1, "Milk", 80m, "литр")
        {
            Category = "Продукты",
            Manufacturer = "Ферма №7",
            Weight = 1.0,
            OriginCountry = "Россия",
            IsOrganic = true,
            QualityGrade = "Высший сорт",
            Volume = 1.0,
            IsRecyclable = true,
            Barcode = "123456789",
            Rating = 4
        };

        PackageItem package = new PackageItem(2, "Apples", 200m, 5)
        {
            Category = "Фрукты",
            Manufacturer = "Green Farm",
            Weight = 2.5,
            PackageWeight = 3.0,
            DeliveryMethod = "Курьер",
            IsFragile = false,
            PackageType = "Картонная коробка",
            ItemsInStock = 50,
            IsBundle = true,
            Barcode = "987654321",
            Rating = 5
        };

        SpecialItem special = new SpecialItem(3, "Chocolate", 150m, DateTime.Now.AddDays(2))
        {
            Category = "Сладости",
            Manufacturer = "Nestle",
            Weight = 0.2,
            ExpirationDate = DateTime.Now.AddDays(5),
            DeliveryMethod = "Почта",
            RequiresCooling = false,
            StorageCondition = "Хранить в сухом месте",
            SpecialCategory = "Премиум",
            LoyaltyPoints = 0,
            IsLimitedEdition = true,
            Barcode = "456123789",
            Rating = 5
        };

        ShoppingCart cart = new ShoppingCart();
        cart.AddItem(single);
        cart.AddItem(package);
        cart.AddItem(special);

        cart.ShowCart();
        cart.DisplayItemTypes();

        Console.WriteLine("--- Демонстрация скидок ---");
        special.ApplyDiscount(0.3m);
        special.ApplyDiscount(0.2m, true); // Перегруженный метод

        Console.WriteLine("--- Расчет стоимости доставки ---");
        Console.WriteLine($"Стоимость доставки яблок: {package.CalculateShippingCost()} руб.");
        Console.WriteLine($"Стоимость доставки шоколада: {special.CalculateShippingCost()} руб.");

        Console.WriteLine("--- Работа с Generic классами ---");
        
        ItemManager<SingleItem> singleItemManager = new ItemManager<SingleItem>();
        singleItemManager.AddItem((SingleItem)single);
        
        ItemManager<Item> allItemManager = new ItemManager<Item>();
        allItemManager.AddItem(single);
        allItemManager.AddItem(package);
        allItemManager.AddItem(special);

        singleItemManager.ProcessItems(item => item.Volume * 1000);

        Inventory<PackageItem> packageInventory = new Inventory<PackageItem>();
        packageInventory.AddToStock(package, 10);
        packageInventory.RemoveFromStock(package, 3);

        Console.WriteLine("--- Новые методы ---");
        single.CheckQuality();
        single.Recycle();
        package.Restock(20);
        special.AwardLoyaltyPoints(50);
        special.CheckExpiration();

        cart.CalculateTotal();

        Console.WriteLine("--- Перегрузка методов ---");
        SingleItem milk = (SingleItem)single;
        milk.ShowLabel();
        milk.ShowLabel(true); // Перегруженный метод

        package.CalculateDiscount(0.1m);
        package.CalculateDiscount(0.1m, "сезонная"); // Перегруженный метод
```
