```polyglot-notebook
public abstract class Component : IInstallable {
    public double Price {get; set;}
    protected string YearOfCreation {get; set;}

    public Component(double price, string yearofcreation) {
        Price = price;
        YearOfCreation = yearofcreation;
    }

    public abstract void Install();

    public virtual void GetInfo() {
        Install();
        Console.WriteLine($"Год создания: {YearOfCreation}\nЦена: {Price}\n".ToString());
    }
}

public interface IInstallable {
    void Install();
}

public class CPU : Component {
    private string Freq {get; set;}

    public CPU(double price, string yearofcreation, string freq) : base(price, yearofcreation) {
        Freq = freq;
    }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Частота: {Freq} GHz");
    }

    public override void Install() {
        Console.WriteLine("Установка в сокет материнской платы");
    }
}

public class RAM : Component {
    private string MemoryType {get; set;}

    public RAM(double price, string yearofcreation, string memorytype) : base(price, yearofcreation) {
        MemoryType = memorytype;
    }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Тип памяти: {MemoryType}");
    }

    public override void Install() {
        Console.WriteLine("Установка в специальные слоты на материнской плате с вариациями: 2; 2 и 4; 1, 2, 3 и 4");
    }
}

public class GPU : Component {
    private string GPUName {get; set;}

    public GPU(double price, string yearofcreation, string gpuname) : base(price, yearofcreation) {
        GPUName = gpuname;
    }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Имя модели: {GPUName}");
    }

    public override void Install() {
        Console.WriteLine("Установка в слот PCI-Express");
    }
}

public class HDD : Component {
    private int StorageVolume {get; set;}

    public HDD(double price, string yearofcreation, int storagevolume) : base(price, yearofcreation) {
        StorageVolume = storagevolume;
    }

    public override void GetInfo() {
        base.GetInfo();
        Console.WriteLine($"Объём накопителя: {StorageVolume} gb");
    }

    public override void Install() {
        Console.WriteLine("Подключение кабеля питания и SATA кабеля, затем крепление в корпусе ПК на винты");
    }
}

public class Computer<T> where T : Component, IInstallable {
    private List<T> _components = new List<T>();

    public void AddComponent(T component) {
        _components.Add(component);
    }

    public void RemoveComponent(T component) {
        var removedComponent = _components.FirstOrDefault<T>(c => c == component);
        _components.Remove(removedComponent);
    }

    public void GetTOtalPrice() {
        double total = 0;

        foreach (T component in _components) {
            total += component.Price;
        }

        Console.WriteLine($"Итоговая цена компонентов: {total}$");
    }
}

CPU newCPU = new CPU(50, "2024", "5.3");
newCPU.GetInfo();

Console.WriteLine("------------------------------------");

RAM newRAM = new RAM(30, "2021", "DDR4");
newRAM.GetInfo();

Console.WriteLine("------------------------------------");

GPU newGPU = new GPU(500, "2023", "RTX 3050");
newGPU.GetInfo();

Console.WriteLine("------------------------------------");

HDD newHDD = new HDD(30, "2015", 512);
newHDD.GetInfo();

Console.WriteLine("------------------------------------");

Computer<Component> newComputer = new Computer<Component>();
newComputer.AddComponent(newCPU);
newComputer.AddComponent(newRAM);
newComputer.AddComponent(newGPU);
newComputer.AddComponent(newHDD);
newComputer.GetTOtalPrice();

Console.WriteLine();

newComputer.RemoveComponent(newGPU);
newComputer.GetTOtalPrice();
```
