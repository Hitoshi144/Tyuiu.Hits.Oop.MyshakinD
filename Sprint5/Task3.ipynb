Создать базовый класс Item в C#, который будет представлять информацию о
товарах, которые могут быть заказаны или возвращены. На основе этого класса
разработать 2-3 производных класса, демонстрирующих принципы наследования и
полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и
методы, а также переопределены некоторые методы базового класса для
демонстрации полиморфизма.

Требования к базовому классу Item:
* Атрибуты: ID товара (ItemId), Название (Name), Цена (Price).
* Методы:
  * GetDetails(): метод для получения детальной информации о товаре.
  * CalculateDiscount(): метод для расчета скидки на товар.
  * ApplyDiscount(decimal discount): метод для применения скидки к цене
товара.

Требования к производным классам:
1. ЕдиничныйТовар (SingleItem): Должен содержать дополнительные атрибуты,
такие как Единица измерения (UnitMeasure). Метод GetDetails() должен быть
переопределен для добавления информации о единице измерения товара.
2. ПакетныйТовар (PackageItem): Должен содержать дополнительные
атрибуты, такие как Количество единиц в пакете (QuantityPerPackage).
Метод CalculateDiscount() должен быть переопределен для учета количества
единиц в пакете при расчете скидки.
3. СпециальныйТовар (SpecialItem) (если требуется третий класс): Должен
содержать дополнительные атрибуты, такие как Дата истечения скидки
(DiscountExpirationDate). Метод ApplyDiscount() должен быть переопределен
для добавления информации о сроке действия скидки.

```
public class ItemManager<T> where T : Item
{
    private List<T> items = new List<T>();

    public void AddItem(T item)
    {
        items.Add(item);
        Console.WriteLine($"Товар '{item.Name}' добавлен в менеджер.");
    }

    public void RemoveItem(int itemId)
    {
        var item = items.FirstOrDefault(i => i.ItemId == itemId);
        if (item != null)
        {
            items.Remove(item);
            Console.WriteLine($"Товар '{item.Name}' удален из менеджера.");
        }
    }

    public void DisplayAllItems()
    {
        Console.WriteLine($"\n--- Все товары в менеджере ({typeof(T).Name}) ---");
        foreach (var item in items)
        {
            item.GetDetails();
        }
    }

    public T FindItemById(int itemId)
    {
        return items.FirstOrDefault(i => i.ItemId == itemId);
    }

    public decimal CalculateTotalValue()
    {
        return items.Sum(item => item.Price);
    }

    public void ProcessItems<U>(Func<T, U> processor)
    {
        foreach (var item in items)
        {
            var result = processor(item);
            Console.WriteLine($"Обработан товар: {item.Name}, результат: {result}");
        }
    }
}

public interface IPerishable
{
    DateTime ExpirationDate { get; set; }
    void CheckExpiration();
    bool IsExpired();
}

public interface IShippable
{
    string DeliveryMethod { get; set; }
    void Ship();
    decimal CalculateShippingCost();
}

public class Inventory<T> where T : Item
{
    private Dictionary<int, int> stock = new Dictionary<int, int>();

    public void AddToStock(T item, int quantity)
    {
        if (stock.ContainsKey(item.ItemId))
            stock[item.ItemId] += quantity;
        else
            stock[item.ItemId] = quantity;

        Console.WriteLine($"Добавлено {quantity} единиц товара '{item.Name}' на склад. Текущий запас: {stock[item.ItemId]}");
    }

    public bool RemoveFromStock(T item, int quantity)
    {
        if (stock.ContainsKey(item.ItemId) && stock[item.ItemId] >= quantity)
        {
            stock[item.ItemId] -= quantity;
            Console.WriteLine($"Изъято {quantity} единиц товара '{item.Name}' со склада. Остаток: {stock[item.ItemId]}");
            return true;
        }
        Console.WriteLine($"Недостаточно товара '{item.Name}' на складе");
        return false;
    }

    public int GetStockQuantity(T item)
    {
        return stock.ContainsKey(item.ItemId) ? stock[item.ItemId] : 0;
    }
}

public class Item
{
    public int ItemId { get; set; }
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Category { get; set; }
    public string Manufacturer { get; set; }
    public double Weight { get; set; }
    public string Barcode { get; set; }
    public DateTime CreatedDate { get; set; }
    public bool IsAvailable { get; set; }
    public int Rating { get; set; }
    
    public string Description { get; set; }
    public string Color { get; set; }
    public string Size { get; set; }
    public int WarrantyMonths { get; set; }
    public List<string> Tags { get; set; }

    public string Material { get; set; }
    public bool IsPopular { get; set; }
    public DateTime LastUpdated { get; set; }

    public delegate void ItemEventHandler(string message);
    public event ItemEventHandler ItemModified;

    public Item(int itemId, string name, decimal price)
    {
        ItemId = itemId;
        Name = name;
        Price = price;
        CreatedDate = DateTime.Now;
        IsAvailable = true;
        Tags = new List<string>();
    }

    public Item() 
    {
        CreatedDate = DateTime.Now;
        IsAvailable = true;
        Tags = new List<string>();
    }

    public virtual void GetDetails()
    {
        Console.WriteLine($"ItemId: {ItemId}");
        Console.WriteLine($"Item Name: {Name}");
        Console.WriteLine($"Item Price: {Price}");
        Console.WriteLine($"Category: {Category}");
        Console.WriteLine($"Manufacturer: {Manufacturer}");
        Console.WriteLine($"Weight: {Weight} кг");
        Console.WriteLine($"Barcode: {Barcode}");
        Console.WriteLine($"Created: {CreatedDate.ToShortDateString()}");
        Console.WriteLine($"Available: {IsAvailable}");
        Console.WriteLine($"Rating: {new string('★', Rating)}{new string('☆', 5 - Rating)}");
        Console.WriteLine($"Description: {Description}");
        Console.WriteLine($"Color: {Color}");
        Console.WriteLine($"Size: {Size}");
        Console.WriteLine($"Warranty: {WarrantyMonths} месяцев");
        Console.WriteLine($"Tags: {string.Join(", ", Tags)}\n");
    }

    public virtual void AddTag(string tag)
    {
        if (!Tags.Contains(tag))
        {
            Tags.Add(tag);
            Console.WriteLine($"Тег '{tag}' добавлен к товару '{Name}'");
        }
    }

    public virtual void RemoveTag(string tag)
    {
        if (Tags.Remove(tag))
        {
            Console.WriteLine($"Тег '{tag}' удален из товара '{Name}'");
        }
    }

    public virtual bool HasTag(string tag)
    {
        return Tags.Contains(tag);
    }

    public virtual void UpdateDescription(string newDescription)
    {
        Description = newDescription;
        Console.WriteLine($"Описание товара '{Name}' обновлено");
    }

    public virtual void ChangeColor(string newColor)
    {
        Color = newColor;
        Console.WriteLine($"Цвет товара '{Name}' изменен на {newColor}");
    }

    public virtual void ExtendWarranty(int additionalMonths)
    {
        WarrantyMonths += additionalMonths;
        Console.WriteLine($"Гарантия на товар '{Name}' продлена на {additionalMonths} месяцев. Общая гарантия: {WarrantyMonths} месяцев");
    }

    public virtual void CalculateDiscount(decimal discount)
    {
        Console.WriteLine($"Price with discount = {Price - (Price * discount)}\n");
    }

    public virtual void CalculateDiscount(decimal discount, string discountName)
    {
        decimal discountedPrice = Price - (Price * discount);
        Console.WriteLine($"Price with {discountName} discount = {discountedPrice}\n");
    }

    public virtual void ApplyDiscount(decimal discount)
    {
        Price -= Price * discount;
    }

    public virtual void UpdateRating(int newRating)
    {
        Rating = Math.Clamp(newRating, 0, 5);
        Console.WriteLine($"Рейтинг товара '{Name}' обновлен: {Rating}");
    }

    public virtual string GetItemType()
    {
        return "Базовый товар";
    }

    public void ChangePrice(decimal newPrice)
    {
        Price = newPrice;
        Console.WriteLine($"Цена товара '{Name}' изменена на {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} (ID: {ItemId}) - {Price} руб. [Гарантия: {WarrantyMonths} мес.]";
    }

    public virtual void UpdateMaterial(string newMaterial)
    {
        Material = newMaterial;
        LastUpdated = DateTime.Now;
        ItemModified?.Invoke($"Материал товара '{Name}' изменен на {newMaterial}");
    }

    public virtual void MarkAsPopular()
    {
        IsPopular = true;
        ItemModified?.Invoke($"Товар '{Name}' отмечен как популярный");
    }

    public virtual void CheckPopularity()
    {
        if (Rating >= 4 && Price < 1000m)
        {
            MarkAsPopular();
        }
    }

    protected virtual void OnItemModified(string message)
    {
        ItemModified?.Invoke(message);
    }
}

public class SingleItem : Item
{
    public string UnitMeasure { get; set; }
    public bool IsOrganic { get; set; }
    public string OriginCountry { get; set; }
    public string QualityGrade { get; set; }
    public double Volume { get; set; }
    public bool IsRecyclable { get; set; }
    
    public DateTime HarvestDate { get; set; }
    public string StorageInstructions { get; set; }
    public bool IsGlutenFree { get; set; }
    public string NutritionalInfo { get; set; }

    public bool IsHandmade { get; set; }
    public string Craftsmanship { get; set; }
    public int ProductionTime { get; set; }

    public delegate void ProductionHandler(string itemName, int days);
    public event ProductionHandler ProductionCompleted;

    public SingleItem(int itemId, string name, decimal price, string unitMeasure)
        : base(itemId, name, price)
    {
        UnitMeasure = unitMeasure;
    }

    public SingleItem() {}

    public override void GetDetails()
    {
        base.GetDetails();
        Console.WriteLine($"Item UnitMeasure: {UnitMeasure}");
        Console.WriteLine($"Organic: {IsOrganic}");
        Console.WriteLine($"Origin Country: {OriginCountry}");
        Console.WriteLine($"Quality Grade: {QualityGrade}");
        Console.WriteLine($"Volume: {Volume}");
        Console.WriteLine($"Recyclable: {IsRecyclable}");
        Console.WriteLine($"Harvest Date: {HarvestDate.ToShortDateString()}");
        Console.WriteLine($"Storage Instructions: {StorageInstructions}");
        Console.WriteLine($"Gluten Free: {IsGlutenFree}");
        Console.WriteLine($"Nutritional Info: {NutritionalInfo}\n");
    }

    public void CheckFreshness()
    {
        TimeSpan shelfLife = DateTime.Now - HarvestDate;
        if (shelfLife.TotalDays > 30)
            Console.WriteLine($"Товар '{Name}' может быть несвежим");
        else
            Console.WriteLine($"Товар '{Name}' свежий");
    }

    public void DisplayNutritionalInfo()
    {
        Console.WriteLine($"Пищевая ценность '{Name}': {NutritionalInfo}");
    }

    public void UpdateStorageInstructions(string instructions)
    {
        StorageInstructions = instructions;
        Console.WriteLine($"Инструкции по хранению для '{Name}' обновлены");
    }

    public bool IsFresh()
    {
        return (DateTime.Now - HarvestDate).TotalDays <= 30;
    }

    public override string GetItemType()
    {
        return "Штучный товар";
    }

    public void CheckQuality()
    {
        Console.WriteLine($"Качество товара '{Name}': {QualityGrade}");
    }

    public void Recycle()
    {
        if (IsRecyclable)
            Console.WriteLine($"Товар '{Name}' отправлен на переработку.");
        else
            Console.WriteLine($"Товар '{Name}' не подлежит переработке.");
    }

    public void ShowLabel()
    {
        Console.WriteLine($"Этикетка: {Name}, {Price} руб. за {UnitMeasure}");
    }

    public void ShowLabel(bool includeOrigin)
    {
        if (includeOrigin)
            Console.WriteLine($"Этикетка: {Name}, {Price} руб. за {UnitMeasure}, Страна: {OriginCountry}");
        else
            ShowLabel();
    }

    public void StartProduction()
    {
        Console.WriteLine($"Начато производство '{Name}'");
        // Имитация производства
        SimulateProduction();
    }

    private void SimulateProduction()
    {
        Task.Delay(1000).ContinueWith(_ => 
        {
            ProductionCompleted?.Invoke(Name, ProductionTime);
            Console.WriteLine($"Производство '{Name}' завершено за {ProductionTime} дней");
        });
    }

    public void DisplayCraftsmanship()
    {
        Console.WriteLine($"Мастерство изготовления '{Name}': {Craftsmanship}");
    }

    public override void UpdateMaterial(string newMaterial)
    {
        base.UpdateMaterial(newMaterial);
        Console.WriteLine($"Материал штучного товара обновлен: {newMaterial}");
    }
}

public interface IVideoReviewable
{
    List<string> VideoReviews { get; set; }
    void AddVideoReview(string videoUrl);
    void PlayVideoReview(int index);
    int VideoReviewsCount { get; }
}

public class PackageItem : Item, IShippable, IVideoReviewable
{
    public int QuantityPerPackage { get; set; }
    public string DeliveryMethod { get; set; }
    public double PackageWeight { get; set; }
    public bool IsFragile { get; set; }
    public string PackageType { get; set; }
    public int ItemsInStock { get; set; }
    public bool IsBundle { get; set; }
    
    public string PackageDimensions { get; set; }
    public bool IsEcoFriendly { get; set; }
    public string AssemblyInstructions { get; set; }
    public int MaxStackHeight { get; set; }

    public bool IsCustomPackage { get; set; }
    public string PackageDesign { get; set; }
    public List<string> PackageFeatures { get; set; }

    private List<string> packageHistory = new List<string>();

    public PackageItem(int itemId, string name, decimal price, int quantityPerPackage)
        : base(itemId, name, price)
    {
        QuantityPerPackage = quantityPerPackage;
        VideoReviews = new List<string>();
        PackageFeatures = new List<string>();
    }

    private List<string> _videoReviews = new List<string>();
    public List<string> VideoReviews 
    { 
        get => _videoReviews;
        set => _videoReviews = value ?? new List<string>();
    }

    void IVideoReviewable.AddVideoReview(string videoUrl)
    {
        if (!string.IsNullOrEmpty(videoUrl))
        {
            _videoReviews.Add(videoUrl);
            Console.WriteLine($"Видеообзор добавлен для упаковки '{Name}'");
        }
    }

    void IVideoReviewable.PlayVideoReview(int index)
    {
        if (index >= 0 && index < _videoReviews.Count)
        {
            Console.WriteLine($"Воспроизведение видеообзора для '{Name}': {_videoReviews[index]}");
        }
        else
        {
            Console.WriteLine("Видеообзор с указанным индексом не найден");
        }
    }

    int IVideoReviewable.VideoReviewsCount => _videoReviews.Count;

    public void AddVideoReview(string videoUrl)
    {
        ((IVideoReviewable)this).AddVideoReview(videoUrl);
    }

    public void PlayVideoReview(int index)
    {
        ((IVideoReviewable)this).PlayVideoReview(index);
    }

    public int VideoReviewsCount => ((IVideoReviewable)this).VideoReviewsCount;

    public PackageItem() 
    {
        VideoReviews = new List<string>();
    }

    public override void GetDetails()
    {
        base.GetDetails();
        Console.WriteLine($"Quantity Per Package: {QuantityPerPackage}");
        Console.WriteLine($"Delivery Method: {DeliveryMethod}");
        Console.WriteLine($"Package Weight: {PackageWeight} кг");
        Console.WriteLine($"Fragile: {IsFragile}");
        Console.WriteLine($"Package Type: {PackageType}");
        Console.WriteLine($"Items In Stock: {ItemsInStock}");
        Console.WriteLine($"Bundle: {IsBundle}");
        Console.WriteLine($"Package Dimensions: {PackageDimensions}");
        Console.WriteLine($"Eco Friendly: {IsEcoFriendly}");
        Console.WriteLine($"Assembly Instructions: {AssemblyInstructions}");
        Console.WriteLine($"Max Stack Height: {MaxStackHeight} единиц");
        Console.WriteLine($"Video Reviews: {VideoReviewsCount}\n");
    }

    public void UpdatePackageDimensions(string dimensions)
    {
        PackageDimensions = dimensions;
        Console.WriteLine($"Размеры упаковки для '{Name}' обновлены: {dimensions}");
    }

    public void DisplayAssemblyInstructions()
    {
        Console.WriteLine($"Инструкции по сборке для '{Name}': {AssemblyInstructions}");
    }

    public bool CanStackWith(PackageItem other)
    {
        return !IsFragile && !other.IsFragile && MaxStackHeight > 0 && other.MaxStackHeight > 0;
    }

    public void EcoFriendlyCheck()
    {
        if (IsEcoFriendly)
            Console.WriteLine($"Упаковка '{Name}' экологически чистая");
        else
            Console.WriteLine($"Упаковка '{Name}' не является экологически чистой");
    }

    public override void CalculateDiscount(decimal discount)
    {
        decimal totalDiscount = (Price * discount) * QuantityPerPackage;
        Console.WriteLine($"Package Discount = {totalDiscount}\n");
    }

    public override void CalculateDiscount(decimal discount, string discountName)
    {
        decimal totalDiscount = (Price * discount) * QuantityPerPackage;
        decimal totalPrice = Price * QuantityPerPackage;
        Console.WriteLine($"Total {discountName} discount for package = {totalDiscount}");
        Console.WriteLine($"Final price = {totalPrice - totalDiscount}\n");
    }

    public void Ship()
    {
        Console.WriteLine($"Товар '{Name}' отправлен методом доставки: {DeliveryMethod}");
    }

    public decimal CalculateShippingCost()
    {
        decimal baseCost = 100m;
        decimal fragileCost = IsFragile ? 50m : 0m;
        decimal weightCost = (decimal)PackageWeight * 10m;
        
        return baseCost + fragileCost + weightCost;
    }

    public void CheckPackaging()
    {
        Console.WriteLine(IsFragile
            ? "Требуется аккуратная упаковка."
            : "Стандартная упаковка подходит.");
    }

    public void Restock(int quantity)
    {
        ItemsInStock += quantity;
        Console.WriteLine($"Пополнен запас товара '{Name}' на {quantity} единиц. Текущий запас: {ItemsInStock}");
    }

    public override string GetItemType()
    {
        return "Упакованный товар";
    }

    public void AddPackageFeature(string feature)
    {
        PackageFeatures.Add(feature);
        packageHistory.Add($"Добавлена особенность упаковки: {feature} - {DateTime.Now}");
        OnItemModified($"Добавлена особенность упаковки для '{Name}': {feature}");
    }

    public void DisplayPackageFeatures()
    {
        Console.WriteLine($"Особенности упаковки '{Name}':");
        foreach (var feature in PackageFeatures)
        {
            Console.WriteLine($" - {feature}");
        }
    }

    public void ShowPackageHistory()
    {
        Console.WriteLine($"История изменений упаковки '{Name}':");
        foreach (var history in packageHistory)
        {
            Console.WriteLine($" - {history}");
        }
    }

    public override void CheckPopularity()
    {
        if (Rating >= 4 && ItemsInStock > 50)
        {
            MarkAsPopular();
            AddPackageFeature("Популярный товар");
        }
    }
}

public class SpecialItem : Item, IPerishable, IShippable
{
    public DateTime DiscountExpirationDate { get; set; }
    public DateTime ExpirationDate { get; set; }
    public string DeliveryMethod { get; set; }
    public bool RequiresCooling { get; set; }
    public string StorageCondition { get; set; }
    public string SpecialCategory { get; set; }
    public int LoyaltyPoints { get; set; }
    public bool IsLimitedEdition { get; set; }
    
    public string LimitedEditionNumber { get; set; }
    public string CertificateAuthenticity { get; set; }
    public bool IsCollectorsItem { get; set; }
    public decimal OriginalValue { get; set; }

    public string InvestmentPotential { get; set; }
    public List<string> PreviousOwners { get; set; }
    public Dictionary<string, decimal> PriceHistory { get; set; }

    public delegate void AuctionHandler(string itemName, decimal currentBid);
    public event AuctionHandler NewBidPlaced;

    public SpecialItem(int itemId, string name, decimal price, DateTime discountExpirationDate)
        : base(itemId, name, price)
    {
        DiscountExpirationDate = discountExpirationDate;
        OriginalValue = price;
        PreviousOwners = new List<string>();
        PriceHistory = new Dictionary<string, decimal>();
        RecordPriceHistory("Начальная цена", price);
    }

    public SpecialItem() 
    {
        OriginalValue = Price;
    }

    public override void GetDetails()
    {
        base.GetDetails();
        Console.WriteLine($"Discount Expiration: {DiscountExpirationDate.ToShortDateString()}");
        Console.WriteLine($"Expiration Date: {ExpirationDate.ToShortDateString()}");
        Console.WriteLine($"Delivery Method: {DeliveryMethod}");
        Console.WriteLine($"Requires Cooling: {RequiresCooling}");
        Console.WriteLine($"Storage Condition: {StorageCondition}");
        Console.WriteLine($"Special Category: {SpecialCategory}");
        Console.WriteLine($"Loyalty Points: {LoyaltyPoints}");
        Console.WriteLine($"Limited Edition: {IsLimitedEdition}");
        Console.WriteLine($"Edition Number: {LimitedEditionNumber}");
        Console.WriteLine($"Certificate: {CertificateAuthenticity}");
        Console.WriteLine($"Collector's Item: {IsCollectorsItem}");
        Console.WriteLine($"Original Value: {OriginalValue} руб.\n");
    }

    public void VerifyAuthenticity()
    {
        if (!string.IsNullOrEmpty(CertificateAuthenticity))
            Console.WriteLine($"Аутентичность '{Name}' подтверждена сертификатом: {CertificateAuthenticity}");
        else
            Console.WriteLine($"Аутентичность '{Name}' не подтверждена");
    }

    public void DisplayCollectorsInfo()
    {
        if (IsCollectorsItem)
        {
            Console.WriteLine($"Коллекционный предмет: {Name}");
            Console.WriteLine($"Оригинальная стоимость: {OriginalValue} руб.");
            Console.WriteLine($"Текущая стоимость: {Price} руб.");
            Console.WriteLine($"Прирост стоимости: {((Price - OriginalValue) / OriginalValue * 100):F1}%");
        }
    }

    public void UpdateEditionNumber(string editionNumber)
    {
        LimitedEditionNumber = editionNumber;
        Console.WriteLine($"Номер издания для '{Name}' обновлен: {editionNumber}");
    }

    public bool IsValuableCollectorsItem()
    {
        return IsCollectorsItem && Price > OriginalValue * 1.5m;
    }

    public override void ApplyDiscount(decimal discount)
    {
        if (DateTime.Now <= DiscountExpirationDate)
        {
            decimal oldPrice = Price;
            base.ApplyDiscount(discount);
            Console.WriteLine($"Скидка {discount:P} применена до {DiscountExpirationDate.ToShortDateString()}");
            Console.WriteLine($"Старая цена: {oldPrice}, Новая цена: {Price}\n");
        }
        else
        {
            Console.WriteLine("Срок действия скидки истек. Скидка не применена.\n");
        }
    }

    public void ApplyDiscount(decimal discount, bool forceApply)
    {
        if (forceApply || DateTime.Now <= DiscountExpirationDate)
        {
            decimal oldPrice = Price;
            base.ApplyDiscount(discount);
            Console.WriteLine($"Скидка {(forceApply ? "принудительно " : "")}применена");
            Console.WriteLine($"Старая цена: {oldPrice}, Новая цена: {Price}\n");
        }
        else
        {
            Console.WriteLine("Срок действия скидки истек. Скидка не применена.\n");
        }
    }

    public void CheckExpiration()
    {
        if (DateTime.Now > ExpirationDate)
            Console.WriteLine($"Товар '{Name}' просрочен!");
        else
            Console.WriteLine($"Товар '{Name}' годен до {ExpirationDate.ToShortDateString()}.");
    }

    public bool IsExpired()
    {
        return DateTime.Now > ExpirationDate;
    }

    public void Ship()
    {
        Console.WriteLine($"'{Name}' отправлен через {DeliveryMethod}. Условия хранения: {StorageCondition}");
    }

    public decimal CalculateShippingCost()
    {
        decimal baseCost = 150m;
        decimal coolingCost = RequiresCooling ? 100m : 0m;
        decimal urgentCost = IsLimitedEdition ? 200m : 0m;
        
        return baseCost + coolingCost + urgentCost;
    }

    public void KeepCool()
    {
        if (RequiresCooling)
            Console.WriteLine($"'{Name}' хранится в холодильнике.");
        else
            Console.WriteLine($"'{Name}' не требует охлаждения.");
    }

    public void AwardLoyaltyPoints(int points)
    {
        LoyaltyPoints += points;
        Console.WriteLine($"Начислено {points} бонусных баллов за покупку '{Name}'. Всего баллов: {LoyaltyPoints}");
    }

    public override string GetItemType()
    {
        return "Специальный товар" + (IsLimitedEdition ? " (Ограниченный тираж)" : "");
    }

    public void AddPreviousOwner(string owner)
    {
        PreviousOwners.Add(owner);
        OnItemModified($"Добавлен предыдущий владелец для '{Name}': {owner}");
    }

    public void RecordPriceHistory(string description, decimal price)
    {
        string key = $"{DateTime.Now:yyyy-MM-dd} - {description}";
        PriceHistory[key] = price;
    }

    public void DisplayPriceHistory()
    {
        Console.WriteLine($"История цен '{Name}':");
        foreach (var history in PriceHistory)
        {
            Console.WriteLine($" - {history.Key}: {history.Value} руб.");
        }
    }

    public void PlaceBid(decimal bidAmount)
    {
        if (bidAmount > Price)
        {
            decimal oldPrice = Price;
            Price = bidAmount;
            RecordPriceHistory("Ставка на аукционе", bidAmount);
            NewBidPlaced?.Invoke(Name, bidAmount);
            Console.WriteLine($"Новая ставка принята: {bidAmount} руб. (предыдущая: {oldPrice} руб.)");
        }
    }

    public override void CheckPopularity()
    {
        if (IsLimitedEdition && LoyaltyPoints > 1000)
        {
            MarkAsPopular();
            InvestmentPotential = "Высокий";
        }
    }
}

public interface IItemService
{
    void ProcessItem(Item item);
    decimal CalculateTotalValue(List<Item> items);
    void GenerateReport(Item item);
}

public class ItemService : IItemService
{
    private readonly ILogger _logger;

    public ItemService(ILogger logger)
    {
        _logger = logger;
    }

    public void ProcessItem(Item item)
    {
        _logger.Log($"Обработка товара: {item.Name}");
        Console.WriteLine($"Обработка товара ID: {item.ItemId}, Название: {item.Name}");
        
        if (item is IPerishable perishable)
        {
            perishable.CheckExpiration();
        }
        
        if (item is IShippable shippable)
        {
            Console.WriteLine($"Стоимость доставки: {shippable.CalculateShippingCost()} руб.");
        }
    }

    public decimal CalculateTotalValue(List<Item> items)
    {
        decimal total = items.Sum(item => item.Price);
        _logger.Log($"Рассчитана общая стоимость: {total} руб.");
        return total;
    }

    public void GenerateReport(Item item)
    {
        _logger.Log($"Генерация отчета для товара: {item.Name}");
        Console.WriteLine($"\n--- ОТЧЕТ ПО ТОВАРУ ---");
        item.GetDetails();
        Console.WriteLine($"Тип товара: {item.GetItemType()}");
        Console.WriteLine($"Статус доступности: {(item.IsAvailable ? "Доступен" : "Не доступен")}");
        Console.WriteLine($"--- КОНЕЦ ОТЧЕТА ---\n");
    }
}

public interface ILogger
{
    void Log(string message);
}

public class ConsoleLogger : ILogger
{
    public void Log(string message)
    {
        Console.WriteLine($"[LOG] {DateTime.Now:HH:mm:ss}: {message}");
    }
}

public class FileLogger : ILogger
{
    public void Log(string message)
    {
        Console.WriteLine($"[LOG] {DateTime.Now:HH:mm:ss}: {message}\n");
    }
}

public class ServiceFactory
{
    public static IItemService CreateItemService(ILogger logger)
    {
        return new ItemService(logger);
    }

    public static ILogger CreateLogger(bool useFileLogger = false, string filePath = null)
    {
        if (useFileLogger && !string.IsNullOrEmpty(filePath))
            return new FileLogger();
        else
            return new ConsoleLogger();
    }
}

public class ItemCollectionManager
{
    private Dictionary<int, Item> itemsDictionary = new Dictionary<int, Item>();
    private HashSet<Item> uniqueItems = new HashSet<Item>();
    private List<Item> popularItems = new List<Item>();

    // Делегаты и события
    public delegate void CollectionEventHandler(string message);
    public event CollectionEventHandler CollectionUpdated;

    public void AddItem(Item item)
    {
        if (!itemsDictionary.ContainsKey(item.ItemId))
        {
            itemsDictionary[item.ItemId] = item;
            uniqueItems.Add(item);
            
            // Подписываемся на событие изменения товара
            item.ItemModified += OnItemModified;
            
            CollectionUpdated?.Invoke($"Товар '{item.Name}' добавлен в коллекцию");
        }
    }

    public void RemoveItem(int itemId)
    {
        if (itemsDictionary.TryGetValue(itemId, out Item item))
        {
            itemsDictionary.Remove(itemId);
            uniqueItems.Remove(item);
            popularItems.Remove(item);
            CollectionUpdated?.Invoke($"Товар '{item.Name}' удален из коллекции");
        }
    }

    public void UpdatePopularItems()
    {
        popularItems = itemsDictionary.Values
            .Where(item => item.IsPopular)
            .OrderByDescending(item => item.Rating)
            .ToList();
        
        CollectionUpdated?.Invoke("Список популярных товаров обновлен");
    }

    public void DisplayAllItems()
    {
        Console.WriteLine("\n=== ВСЕ ТОВАРЫ В КОЛЛЕКЦИИ ===");
        foreach (var item in itemsDictionary.Values)
        {
            item.GetDetails();
        }
    }

    public void DisplayPopularItems()
    {
        Console.WriteLine("\n=== ПОПУЛЯРНЫЕ ТОВАРЫ ===");
        foreach (var item in popularItems)
        {
            item.GetDetails();
        }
    }

    public Item FindItemById(int itemId)
    {
        return itemsDictionary.GetValueOrDefault(itemId);
    }

    public List<Item> FindItemsByTag(string tag)
    {
        return itemsDictionary.Values
            .Where(item => item.HasTag(tag))
            .ToList();
    }

    private void OnItemModified(string message)
    {
        Console.WriteLine($"[Событие товара]: {message}");
    }
}

ILogger logger = ServiceFactory.CreateLogger();
        
IItemService itemService = ServiceFactory.CreateItemService(logger);

SingleItem milk = new SingleItem(1, "Organic Milk", 120m, "литр")
{
    Category = "Молочные продукты",
    Manufacturer = "Эко-Ферма",
    Weight = 1.0,
    OriginCountry = "Россия",
    IsOrganic = true,
    QualityGrade = "Высший сорт",
    Volume = 1.0,
    IsRecyclable = true,
    Barcode = "123456789",
    Rating = 5,
    Description = "Свежее органическое молоко",
    Color = "Белый",
    Size = "1л",
    WarrantyMonths = 3,
    HarvestDate = DateTime.Now.AddDays(-5),
    StorageInstructions = "Хранить при +4°C",
    IsGlutenFree = true,
    NutritionalInfo = "Белки: 3.2г, Жиры: 3.5г, Углеводы: 4.7г"
};

PackageItem apples = new PackageItem(2, "Red Apples", 350m, 10)
{
    Category = "Фрукты",
    Manufacturer = "Apple Garden",
    Weight = 5.0,
    PackageWeight = 5.5,
    DeliveryMethod = "Курьер",
    IsFragile = false,
    PackageType = "Деревянный ящик",
    ItemsInStock = 100,
    IsBundle = true,
    Barcode = "987654321",
    Rating = 4,
    Description = "Свежие красные яблоки",
    Color = "Красный",
    Size = "Средний",
    WarrantyMonths = 1,
    PackageDimensions = "30x20x15 см",
    IsEcoFriendly = true,
    AssemblyInstructions = "Не требует сборки",
    MaxStackHeight = 5
};

SpecialItem wine = new SpecialItem(3, "Vintage Wine", 2500m, DateTime.Now.AddDays(7))
{
    Category = "Алкоголь",
    Manufacturer = "Chateau Bordeaux",
    Weight = 1.5,
    ExpirationDate = DateTime.Now.AddYears(10),
    DeliveryMethod = "Спецдоставка",
    RequiresCooling = true,
    StorageCondition = "Хранить горизонтально при +12°C",
    SpecialCategory = "Премиум алкоголь",
    LoyaltyPoints = 500,
    IsLimitedEdition = true,
    Barcode = "456123789",
    Rating = 5,
    Description = "Выдержанное красное вино",
    Color = "Красный",
    Size = "0.75л",
    WarrantyMonths = 0,
    LimitedEditionNumber = "LE-2024-001",
    CertificateAuthenticity = "Сертификат подлинности №12345",
    IsCollectorsItem = true,
    OriginalValue = 2000m
};

milk.AddTag("organic");
milk.AddTag("dairy");
milk.AddTag("fresh");
        
apples.AddTag("fruits");
apples.AddTag("healthy");
apples.AddTag("eco");
        
wine.AddTag("alcohol");
wine.AddTag("premium");
wine.AddTag("collectors");

Console.WriteLine("=== ДЕМОНСТРАЦИЯ НОВЫХ МЕТОДОВ ===");
        
milk.CheckFreshness();
milk.DisplayNutritionalInfo();
milk.UpdateStorageInstructions("Хранить при +2°C до +6°C");
        
apples.UpdatePackageDimensions("35x25x20 см");
apples.DisplayAssemblyInstructions();
apples.EcoFriendlyCheck();
        
wine.VerifyAuthenticity();
wine.DisplayCollectorsInfo();
wine.UpdateEditionNumber("LE-2024-001-B");

Console.WriteLine("\n=== ЯВНАЯ РЕАЛИЗАЦИЯ ИНТЕРФЕЙСА ===");
apples.AddVideoReview("https://youtube.com/watch?v=apples_review1");
apples.AddVideoReview("https://youtube.com/watch?v=apples_review2");
apples.PlayVideoReview(0);
Console.WriteLine($"Количество видеообзоров: {apples.VideoReviewsCount}");

Console.WriteLine("\n=== УПРАВЛЕНИЕ ЗАВИСИМОСТЯМИ ===");
itemService.ProcessItem(milk);
itemService.ProcessItem(apples);
itemService.ProcessItem(wine);

List<Item> allItems = new List<Item> { milk, apples, wine };
decimal totalValue = itemService.CalculateTotalValue(allItems);
Console.WriteLine($"Общая стоимость всех товаров: {totalValue} руб.");

itemService.GenerateReport(wine);

Console.WriteLine("\n=== РАБОТА С GENERIC КЛАССАМИ ===");
ItemManager<SingleItem> singleManager = new ItemManager<SingleItem>();
singleManager.AddItem(milk);
        
ItemManager<PackageItem> packageManager = new ItemManager<PackageItem>();
packageManager.AddItem(apples);

Inventory<PackageItem> inventory = new Inventory<PackageItem>();
inventory.AddToStock(apples, 50);
inventory.RemoveFromStock(apples, 10);

Console.WriteLine("\n=== ПЕРЕГРУЖЕННЫЕ МЕТОДЫ ===");
milk.ShowLabel();
milk.ShowLabel(true);
        
apples.CalculateDiscount(0.1m);
apples.CalculateDiscount(0.15m, "праздничная");

Console.WriteLine($"\nВинто является ценным коллекционным предметом: {wine.IsValuableCollectorsItem()}");
Console.WriteLine($"Молоко свежее: {milk.IsFresh()}");
Console.WriteLine($"Яблоки можно штабелировать: {apples.CanStackWith(apples)}");

Console.WriteLine("======================================");

var collectionManager = new ItemCollectionManager();
collectionManager.CollectionUpdated += (message) => 
    Console.WriteLine($"[Менеджер коллекций]: {message}");

var handmadePottery = new SingleItem(4, "Handmade Pottery", 2500m, "штука")
{
    Material = "Глина",
    IsHandmade = true,
    Craftsmanship = "Ручная работа",
    ProductionTime = 14,
    Category = "Ремесла",
    Rating = 5
};

var luxuryPackage = new PackageItem(5, "Luxury Gift Set", 5000m, 1)
{
    Material = "Картон премиум",
    IsCustomPackage = true,
    PackageDesign = "Подарочная упаковка",
    Category = "Подарки",
    Rating = 5
};

var antiqueItem = new SpecialItem(6, "Antique Vase", 15000m, DateTime.Now.AddDays(30))
{
    Material = "Фарфор",
    InvestmentPotential = "Средний",
    IsLimitedEdition = true,
    IsCollectorsItem = true,
    Rating = 5
};

collectionManager.AddItem(handmadePottery);
collectionManager.AddItem(luxuryPackage);
collectionManager.AddItem(antiqueItem);

Console.WriteLine("=== ДЕМОНСТРАЦИЯ НОВЫХ ФУНКЦИОНАЛЬНОСТЕЙ ===");

handmadePottery.ProductionCompleted += (name, days) =>
    Console.WriteLine($"✅ Производство завершено: {name} за {days} дней");

antiqueItem.NewBidPlaced += (name, bid) =>
    Console.WriteLine($"🔥 Новая ставка на аукционе: {name} - {bid} руб.");

handmadePottery.StartProduction();

luxuryPackage.AddPackageFeature("Золотое тиснение");
luxuryPackage.AddPackageFeature("Шелковая лента");
luxuryPackage.DisplayPackageFeatures();

antiqueItem.PlaceBid(16000m);
antiqueItem.PlaceBid(17000m);
antiqueItem.DisplayPriceHistory();

antiqueItem.AddPreviousOwner("Частный коллекционер");
antiqueItem.AddPreviousOwner("Музей искусств");

handmadePottery.CheckPopularity();
luxuryPackage.CheckPopularity();
antiqueItem.CheckPopularity();

collectionManager.UpdatePopularItems();

collectionManager.DisplayAllItems();
collectionManager.DisplayPopularItems();

handmadePottery.AddTag("handmade");
handmadePottery.AddTag("craft");
var handmadeItems = collectionManager.FindItemsByTag("handmade");
Console.WriteLine($"\nНайдено товаров ручной работы: {handmadeItems.Count}");

handmadePottery.UpdateMaterial("Керамика ручной работы");
luxuryPackage.UpdateMaterial("Экокожа");

Console.WriteLine("\n=== РАБОТА С РАЗЛИЧНЫМИ КОЛЛЕКЦИЯМИ ===");
        
var uniqueItemsSet = new HashSet<Item> { handmadePottery, luxuryPackage, antiqueItem };
Console.WriteLine($"Уникальных товаров в HashSet: {uniqueItemsSet.Count}");

var itemsDict = new Dictionary<string, Item>
{
    ["pottery"] = handmadePottery,
    ["gift"] = luxuryPackage,
    ["antique"] = antiqueItem
};

foreach (var kvp in itemsDict)
{
    Console.WriteLine($"Ключ: {kvp.Key}, Товар: {kvp.Value.Name}");
}

var expensiveItems = itemsDict.Values
    .Where(item => item.Price > 3000m)
    .OrderByDescending(item => item.Price)
    .ToList();

Console.WriteLine("\n=== ДОРОГИЕ ТОВАРЫ ( > 3000 руб.) ===");
foreach (var item in expensiveItems)
{
    Console.WriteLine($"{item.Name}: {item.Price} руб.");
}
```
