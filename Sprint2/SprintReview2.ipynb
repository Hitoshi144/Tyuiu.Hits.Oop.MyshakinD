```polyglot-notebook
public class Employee {
    protected string Name {get; set;}
    protected string Position {get; set;}
    protected double Salary {get; set;}

    public Employee(string name, string position, double salary) {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo() {
        Console.WriteLine($"Имя сотрудника: {Name}\nДолжность: {Position}\nЗарплата: {Salary}");
    }
}

public class Manager : Employee {
    private string Departament {get; set;}

    public Manager(string name, string position, string departament, double salary) : base(name, position, salary) {
        Departament = departament;
    }

    public override void PrintInfo() {
        base.PrintInfo();
        Console.WriteLine($"Департамент: {Departament}");
    }
}

public interface IReportable {
    string GenerateReport();
}

public interface ITeamLeader {
    List<Employee> GetTeamMembers();
}

public class ProjectManager : Manager, IReportable, ITeamLeader {
    private string ProjectName {get; set;}
    private List<Employee> TeamMembers {get; set;}
    private string TeamName {get; set;}

    public ProjectManager(string name, string position, string departament, string projectname, double salary,
    Employee member, string teamname
    ) : base(name, position, departament, salary) {
        ProjectName = projectname;
        TeamMembers = new List<Employee>();
        TeamMembers.Add(member);
        TeamName = teamname;
    }

    public override void PrintInfo() {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
    }

    public string GenerateReport() {
        string report = $"Название команды: {TeamName}\nТекущий проект: {ProjectName}";
        return report;
    }

    public List<Employee> GetTeamMembers() {
        return TeamMembers;
    }

    public void AddMember(Employee member) {
        TeamMembers.Add(member);
    }

    public void RemoveMember(Employee member) {
        var removedMember = TeamMembers.FirstOrDefault<Employee>(m => m == member);

        TeamMembers.Remove(removedMember);
    }
}

Employee newEmployee = new Employee("Сергей", "Разнорабочий", 42151.12);
newEmployee.PrintInfo();

Console.WriteLine("----------");

Manager newManager = new Manager("Олег", "Менеджер", "Внутренних дел", 514161.41);
newManager.PrintInfo();

Console.WriteLine("----------");

ProjectManager newProjectManager = new ProjectManager("Владимир", "Проектный менеджер", "Внутренниз дел", "AIVision", 66246264.61, newManager, "Звезда");
newProjectManager.PrintInfo();

Console.WriteLine("----------");

Console.WriteLine(newProjectManager.GenerateReport());
Console.WriteLine();
foreach (var member in newProjectManager.GetTeamMembers()) {
    member.PrintInfo();
}

Console.WriteLine("----------");

newProjectManager.AddMember(newEmployee);
foreach (var member in newProjectManager.GetTeamMembers()) {
    member.PrintInfo();
    Console.WriteLine();
}

Console.WriteLine("----------");

newProjectManager.RemoveMember(newEmployee);
foreach (var member in newProjectManager.GetTeamMembers()) {
    member.PrintInfo();
}
```
