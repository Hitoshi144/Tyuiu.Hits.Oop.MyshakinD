```polyglot-notebook
public class Plant {
    protected string Name {get; set;}
    protected string Type {get; set;}
    protected float Height {get; set;}

    public Plant(string name, string type, float height) {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom() {
        Console.WriteLine("Растение цветёт");
    }

    public void Grow(float height) {
        Height += height;
        Console.WriteLine($"Новая высота: {Height}");
    }
}

public class Flower : Plant {
    private string Color {get; set;}
    private int PetalsCount {get; set;}

    public Flower(
        string color, 
        int petalscount,
        string name,
        string type,
        float height
        ) : base(name, type, height) {
            Color = color;
            PetalsCount = petalscount;
    }
    
    public override void Bloom() {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Fade() {
        Console.WriteLine("Цветок начинает увядать");
    }
}

public class Tree : Plant {
    private int Age {get; set;}
    private string LeavesType {get; set;}

    public Tree(
        int age,
        string leavestype,
        string name,
        string type,
        float height
    ) : base(name, type, height) {
        Age = age;
        LeavesType = leavestype;
    }

    public override void Bloom() {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void GrowUp(int ages) {
        Age += ages;
        Console.WriteLine($"Новый возраст дерева: {Age}");
    }
}

Plant newPlant = new Plant("Роза", "Цветок", 1.1f);
newPlant.Bloom();
newPlant.Grow(0.1f);

Console.WriteLine();

Flower newFlower = new Flower("Ярко-красный", 9, "Ниневия", "Цветок", 2.4f);
newFlower.Bloom();
newFlower.Fade();

Console.WriteLine();

Tree newTree = new Tree(128, "Хвойные", "Ель", "Дерево", 77f);
newTree.Bloom();
newTree.GrowUp(1);
```
